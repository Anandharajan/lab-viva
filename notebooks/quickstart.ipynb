## 2. Run an adaptive viva (answers simulated here; in class the examiner marks right/wrong)

````python
import random
from lab_viva import api
q = api.start_viva("S001")
while not q["done"]:
    print(f"[{q['bloom']}] {q['question']}")
    q = api.record("S001", correct=random.random() < 0.7)
print(f"ability {q['theta']:+.2f} (SE {q['se']:.2f}) after {len(q['asked'])} questions")
````
